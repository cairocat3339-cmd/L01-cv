````python
# Your Phase 1 solution to the problem goes here.
import pickle
with open('data/minst.pkl', 'rb') as f:
    data = pickle.load(f)
X_train, y_train = data['X_train'], data['y_train']
X_test, y_test = data['X_test'], data['y_test']
````

````python
import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000)
for i in range(5): 
    print(y_test[i],'\n') # Print the label
    print(X_test[i],'\n') # Print the matrix of values
    plt.contourf(np.rot90(X_test[i].transpose())) # Make a contour plot of the matrix values
    plt.show()
````

````python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("One training image:", X_train[0].shape)
print("One test image:", X_test[0].shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

X_train_flat = X_train.reshape((len(X_train), 784))
X_test_flat = X_test.reshape((len(X_test), 784))
````

There are 60,000 training images and 10,000 test images. Each image is a 28 × 28 matrix. The label arrays contain one digit label per image. Reshaping gives one row per image and 784 pixel features per row.

````python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

X_fit, X_valid, y_fit, y_valid = train_test_split(
    X_train_flat, y_train, test_size=0.20, random_state=65
)

ks = [1, 3, 5]
mistakes = []

for k in ks:
    model = KNeighborsClassifier(n_neighbors=k).fit(X_fit, y_fit)
    y_hat = model.predict(X_valid)
    mistakes.append(sum(y_valid != y_hat))
    print("k:", k, "Validation accuracy:", sum(y_valid == y_hat) / len(y_valid))

best_k = ks[np.argmin(mistakes)]
print("Selected k:", best_k)
````

We compared k = 1, 3, and 5 and selected the value with the fewest validation mistakes. This is the best of the values tested, rather than a guaranteed optimal value. We then fit that model on all the training images and evaluated it on the test images.

````python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_valid, y_hat)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
````

The most common mistakes involve mistaking 5 for 3, 8 for 6, and 7 for 1. There are no significant patterns shown in the confusion matrix, though certain rows (i.e. 0, 1, 6, and 9) have fewer errors than other entries (i.e. 5 and 8), indicating that shape complexity has a direct relationship with prediction errors.
